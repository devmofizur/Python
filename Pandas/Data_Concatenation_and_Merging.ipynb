```python
import pandas as pd
import numpy as np
```

```python
data1 = pd.DataFrame({
    'city': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'total_house': [350000, 420000, 280000, 310000, 250000],
    'total_hotels': [300, 450, 200, 280, 150]
})

data2 = pd.DataFrame({
    'city': ['Bhola','Shylhet', 'Dinajpur','Feni','Jessore'],
    'total_house': [20000, 38000, 15000, 40000, 9000],
    'total_hotels': [100, 350, 80, 400, 60]
})
```

```python
df = pd.concat([data1,data2])
df
```

```python
df = pd.concat([data1,data2], ignore_index=True)
df
```

```python
df = pd.concat([data1,data2], keys = ['US', 'BD'])
df
```

```python
df.loc['BD']
```

```python
df = pd.concat([data1,data2], keys = ['US', 'BD'], axis =1)
df
```

```python
data1 = pd.DataFrame({
    'city': ['London', 'Paris', 'Berlin', 'Rome', 'Madrid'],
    'average_temperature': [12, 15, 10, 18, 16],
    'average_humidity': [75, 70, 80, 65, 60]
})

data2 = pd.DataFrame({
    'city': ['Cairo', 'Dubai', 'Mumbai', 'Sydney', 'Tokyo'],
    'average_temperature': [25, 30, 28, 15, 19],
    'average_humidity': [50, 40, 70, 70, 68]
})
```

```python
df3 = pd.merge(data1,data2, on = 'average_temperature') #SQL JOIN
df3
```

```python
data1 = pd.DataFrame({
    'item': ['Apple', 'Banana', 'Orange', 'Grape', 'Kiwi'],
    'price': [1.2, 0.5, 0.8, 2.0, 1.5],
    'quantity': [100, 250, 180, 70, 120]
})

data2 = pd.DataFrame({
    'item': ['Milk', 'Bread', 'Eggs', 'Grape', 'Butter'],
    'price': [3.0, 2.5, 4.0, 7.0, 5.0],
    'quantity': [50, 80, 60, 30, 40]
})
```

```python
df4 = pd.merge(data1,data2,  how = 'outer')
df4
```

```python
df4 = pd.merge(data1,data2, how = 'left')
df4
```

```python
df4 = pd.merge(data1, data2,  how='right')
df4
```

```python
df4 = pd.merge(data1, data2, how='inner')
df4
```

```python
df4 = pd.merge(data1,data2, on='item', how = 'outer')
df4
```

```python
df4 = pd.merge(data1,data2, on='item', how = 'left')
df4
```

```python
df4 = pd.merge(data1,data2, on='item', how = 'right')
df4
```

```python
df4 = pd.merge(data1,data2, on='item', how = 'inner')
df4
```
