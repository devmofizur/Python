```python
from google.colab import drive
import shutil
import os

drive.mount('/content/drive')

source_path = '/content/drive/MyDrive/Colab Notebooks/ML Learning/linechart.xlsx'
destination_path = '/content/linechart.xlsx'

if os.path.exists(source_path):
    shutil.copy(source_path, destination_path)
    print(f"Successfully copied '{source_path}' to '{destination_path}'")
else:
    print(f"Error: Source file '{source_path}' not found.")
```

```python
import pandas as pd
df = pd.read_excel('linechart.xlsx')
df.head()
```

## Line Chart



```python
from numpy import left_shift
from matplotlib import pyplot as plt

plt.figure(figsize=(6,5))
plt.plot(df['Quarter'],df['Fridge'], label='Fridge')
plt.plot(df['Quarter'],df['Dishwasher'], label='Dishwasher')
plt.plot(df['Quarter'],df['Washing Machine'], label='Washing Machine')
plt.title('Product sales')
plt.xlabel('Quarter')
plt.ylabel('Usage')
plt.legend(loc='upper left')
plt.grid(True)
plt.show()
```

## Pie Chart

```python
total_sales = df[['Fridge', 'Dishwasher', 'Washing Machine']].sum()
total_sales
```

```python
total_sales.index
```

```python
plt.pie(total_sales, labels=total_sales.index, autopct='%1.2f%%', explode=(0.05,0,0), shadow=True, startangle=90)
plt.title('Total Sales by Product')
plt.show()
```

## Bar Chart

```python
df.plot(kind='bar', x = 'Quarter')
plt.xlabel('Quarter')
plt.ylabel('Usage')
plt.title('Product Sales')
plt.xticks(rotation=45)
plt.legend(loc='upper left')

plt.show()
```
